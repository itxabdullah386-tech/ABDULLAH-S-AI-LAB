```python
import random

# -----------------------------
# Problem Data
# -----------------------------
weights = [2, 3, 4, 5, 6]
values = [10, 20, 30, 40, 50]

capacity = 10

population_size = 10
generations = 50
mutation_rate = 0.1


# -----------------------------
# Create Initial Population
# -----------------------------
def create_population():
    population = []

    for _ in range(population_size):
        chromosome = [
            random.randint(0, 1)
            for _ in range(len(weights))
        ]
        population.append(chromosome)

    return population


# -----------------------------
# Fitness Function
# -----------------------------
def fitness(chromosome):
    total_weight = 0
    total_value = 0

    for i in range(len(chromosome)):
        if chromosome[i] == 1:
            total_weight += weights[i]
            total_value += values[i]

    # If capacity is exceeded
    if total_weight > capacity:
        return 0

    return total_value


# -----------------------------
# Selection
# -----------------------------
def selection(population):
    population.sort(key=fitness, reverse=True)

    # Select best half
    return population[:population_size // 2]


# -----------------------------
# Crossover
# -----------------------------
def crossover(parent1, parent2):
    point = random.randint(1, len(weights) - 1)

    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]

    return child1, child2


# -----------------------------
# Mutation
# -----------------------------
def mutation(chromosome):
    for i in range(len(chromosome)):
        if random.random() < mutation_rate:
            chromosome[i] = 1 - chromosome[i]

    return chromosome


# -----------------------------
# Genetic Algorithm
# -----------------------------
def genetic_algorithm():
    population = create_population()

    for generation in range(generations):

        parents = selection(population)

        new_population = parents.copy()

        while len(new_population) < population_size:

            parent1 = random.choice(parents)
            parent2 = random.choice(parents)

            child1, child2 = crossover(parent1, parent2)

            child1 = mutation(child1)
            child2 = mutation(child2)

            new_population.append(child1)

            if len(new_population) < population_size:
                new_population.append(child2)

        population = new_population

    # Find best solution
    best_solution = max(population, key=fitness)

    return best_solution


# -----------------------------
# Run GA
# -----------------------------
best = genetic_algorithm()

total_weight = sum(
    weights[i] for i in range(len(weights))
    if best[i] == 1
)

total_value = sum(
    values[i] for i in range(len(values))
    if best[i] == 1
)

print("Best Chromosome:", best)
print("Total Weight:", total_weight)
print("Total Value:", total_value)
```

```python
import random

# -----------------------------
# Problem Data
# -----------------------------
classes = ["Math", "Physics", "Programming", "English"]

teachers = ["T1", "T2", "T1", "T3"]

rooms = ["Room1", "Room2"]

times = ["9AM", "10AM", "11AM"]

population_size = 10
generations = 50
mutation_rate = 0.1


# -----------------------------
# Create Initial Population
# -----------------------------
def create_population():

    population = []

    for _ in range(population_size):

        chromosome = []

        for i in range(len(classes)):

            time = random.choice(times)
            room = random.choice(rooms)

            chromosome.append((time, room))

        population.append(chromosome)

    return population


# -----------------------------
# Fitness Function
# -----------------------------
def fitness(chromosome):

    conflicts = 0

    # Check every pair of classes
    for i in range(len(classes)):

        for j in range(i + 1, len(classes)):

            time1, room1 = chromosome[i]
            time2, room2 = chromosome[j]

            # Same time + same room
            if time1 == time2 and room1 == room2:
                conflicts += 1

            # Same teacher + same time
            if time1 == time2 and teachers[i] == teachers[j]:
                conflicts += 1

    # Higher fitness = better timetable
    return 1 / (1 + conflicts)


# -----------------------------
# Selection
# -----------------------------
def selection(population):

    population.sort(key=fitness, reverse=True)

    return population[:population_size // 2]


# -----------------------------
# Crossover
# -----------------------------
def crossover(parent1, parent2):

    point = random.randint(1, len(classes) - 1)

    child1 = parent1[:point] + parent2[point:]

    child2 = parent2[:point] + parent1[point:]

    return child1, child2


# -----------------------------
# Mutation
# -----------------------------
def mutation(chromosome):

    for i in range(len(chromosome)):

        if random.random() < mutation_rate:

            new_time = random.choice(times)
            new_room = random.choice(rooms)

            chromosome[i] = (new_time, new_room)

    return chromosome


# -----------------------------
# Genetic Algorithm
# -----------------------------
def genetic_algorithm():

    population = create_population()

    for generation in range(generations):

        parents = selection(population)

        new_population = parents.copy()

        while len(new_population) < population_size:

            parent1 = random.choice(parents)
            parent2 = random.choice(parents)

            child1, child2 = crossover(parent1, parent2)

            child1 = mutation(child1)
            child2 = mutation(child2)

            new_population.append(child1)

            if len(new_population) < population_size:
                new_population.append(child2)

        population = new_population

    best_solution = max(population, key=fitness)

    return best_solution


# -----------------------------
# Run GA
# -----------------------------
best = genetic_algorithm()

print("\nBest Timetable:\n")

for i in range(len(classes)):

    time, room = best[i]

    print(
        classes[i],
        "->",
        time,
        "->",
        room,
        "-> Teacher:",
        teachers[i]
    )

print("\nFitness:", fitness(best))
```
