```python
df['discounted_price'] = df['discounted_price'].str.replace('₹', '', regex=False).str.replace(',', '', regex=False).astype(float)
df['actual_price'] = df['actual_price'].str.replace('₹', '', regex=False).str.replace(',', '', regex=False).astype(float)

df['discount_percentage'] = df['discount_percentage'].str.replace('%', '', regex=False).astype(float)

df[['discounted_price', 'actual_price', 'discount_percentage']].head()
print(df.dtypes)
```

```python
le = LabelEncoder()

df['product_id'] = le.fit_transform(df['product_id'])
df['category'] = le.fit_transform(df['category'])

df.head()
```

```python
columns_to_drop = [
    'product_name', 'rating', 'rating_count', 'about_product', 
    'user_id', 'user_name', 'review_id', 'review_title', 
    'img_link', 'product_link'
]

df_cleaned = df.drop(columns=columns_to_drop)

df_cleaned.head()
```

```python
df_cleaned = df_cleaned.drop(columns=['review_content'])
df_cleaned
```

```python
df_cleaned['final_price'] = df_cleaned['actual_price'] - df_cleaned['discounted_price']
df_cleaned
```

```python
x = df_cleaned[["product_id", "category",	"discounted_price", "actual_price", "discount_percentage"]]
y = df_cleaned[["final_price"]]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_predict = model.predict(x_test)
model.predict([[201,172,2280.0,3045.0,25.0]])
```

```python
mse = mean_squared_error(y_test, y_predict)
score = r2_score(y_test, y_predict)
```

```python
print(f"Mean Squared Value: {mse:.2f}")
print(f"R2 Score is: {score:.2f}")
```

```python
y_test_flat = y_test.values.flatten()
y_predict_flat = y_predict.flatten()

mean_absolute_err = np.mean(np.abs(y_predict_flat - y_test_flat))

mean_actual_val = np.mean(y_test_flat)

error_percentage = (mean_absolute_err / mean_actual_val) * 100

accuracy = 100 - error_percentage

print(f"Mean Absolute Error: {mean_absolute_err:.2f}")
print(f"Mean Actual Value: {mean_actual_val:.2f}")
print(f"Approximate Accuracy: {accuracy:.2f}%")
```
